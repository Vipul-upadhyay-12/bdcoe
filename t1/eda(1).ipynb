```python
df["popularity"].hist(bins=80, color="pink", alpha = .1)
plt.xlabel("popularity")
plt.ylabel("number of tracks")
plt.title("Popularity Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.select_dtypes(include="number").corr(), annot=False, cmap="coolwarm")
plt.title("Correlation between numeric features")
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.scatter(df["energy"], df["loudness"], alpha=0.1, s=5)
plt.xlabel("energy")
plt.ylabel("loudness")
plt.title("Energy vs Loudness")
plt.show()
```

```python
top_genres = df.groupby("track_genre")["popularity"].mean().sort_values(ascending=False).head(10)
top_genres
```

```python
top_genres.plot(kind="bar", figsize=(5,5), color="skyblue")
plt.ylabel("average popularity")
plt.title("Top 10 Genres by Average Popularity")
plt.show()
```
